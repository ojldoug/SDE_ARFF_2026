```python
import numpy as np

from gillespie.sir import (
    SIRG,
    histogram_data_SIRG,
    plot_histogram_SIRG
)

%load_ext autoreload
%autoreload 2
```

```python
# notebook parameters
n_dimensions = 2
random_seed = 1

# SIR model paramters
k1, k2, k3 = 1, 1, 0
N = 1024

# training data parameters
n_trajectories = 250
trajectory_time = 4
time_step = 1e-3
n_skip_steps = 10
```

```python
sirg = SIRG(N=N, k1=k1, k2=k2, k3=0, random_state=(random_seed))
x_data, y_data, step_sizes = sirg.sample_data_SIRG(n_trajectories, trajectory_time, time_step, n_skip_steps)
r_data = y_data - x_data
```

```python
print(x_data.shape, r_data.shape, step_sizes.shape)
```

```python
diff_type = "diagonal"
np.savez(f"training_data/ex5.npz", x_data=np.array(x_data), r_data=np.array(r_data), step_sizes=step_sizes, diff_type=diff_type)
```
