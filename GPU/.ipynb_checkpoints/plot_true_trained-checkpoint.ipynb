```python
import numpy as np
import dill
import jax
import jax.numpy as jnp
from matplotlib import pyplot as plt
from matplotlib.colors import Normalize
import matplotlib.ticker as mticker
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
ex_name = "ex6"

# load data with sampling
training_data = np.load(f"training_data/{ex_name}.npz")

x = training_data["x_data"]
r = training_data["r_data"]
h = training_data["step_sizes"]
diff_type = str(training_data["diff_type"])
print(x.shape, r.shape, h.shape)

if x.ndim == 3:
    # subsample data
    sample_rate = 1000
    x = x[:, :, ::sample_rate].reshape(-1, x.shape[1])
    r = r.reshape(r.shape[0], r.shape[1], r.shape[2] // sample_rate, sample_rate).sum(axis=3).reshape(-1, r.shape[1])
    h = h.reshape(h.shape[0], h.shape[1] // sample_rate, sample_rate).sum(axis=2)
    
print(x.shape, r.shape, h.shape)

yD = r.shape[1]
xD = x.shape[1]
```

```python
# load true functions
with open(f"true_functions/{ex_name}.pkl", "rb") as f:
    true_functions = dill.load(f)

true_drift = true_functions["drift"]
true_diffusion = true_functions["diffusion"]

# get true and ARFF param
from lib.lib_ARFF_r import Functions as ARFF_Functions
from lib.lib_Adam_FF import Functions as Adam_Functions_FF
from lib.lib_Adam_tanh import Functions as Adam_Functions_tanh

ARFF_param = np.load(f"saved_results/NN_param/{ex_name}_ARFF.npz", allow_pickle=True)
Adam_param_FF = np.load(f"saved_results/NN_param/{ex_name}_Adam.npz", allow_pickle=True)
Adam_param_tanh = np.load(f"saved_results/NN_param/{ex_name}_tanh_Adam.npz", allow_pickle=True)
Adam_param_deep = np.load(f"saved_results/NN_param/{ex_name}_deep_Adam.npz", allow_pickle=True)

def to_jax_dict(d):
    """Convert all arrays in a {str: array} dict to jax arrays."""
    return {k: jnp.asarray(v) for k, v in d.items()}

# Convert only the relevant sub-dicts
ARFF_drift_param = {k: jnp.asarray(v) for k, v in ARFF_param["drift_param"].item().items()}
ARFF_diff_param  = {k: jnp.asarray(v) for k, v in ARFF_param["diff_param"].item().items()}

Adam_drift_param_FF = {k: jnp.asarray(v) for k, v in Adam_param_FF["drift_param"].item().items()}
Adam_diff_param_FF  = {k: jnp.asarray(v) for k, v in Adam_param_FF["diff_param"].item().items()}

Adam_drift_param_tanh = {k: jnp.asarray(v) for k, v in Adam_param_tanh["drift_param"].item().items()}
Adam_diff_param_tanh  = {k: jnp.asarray(v) for k, v in Adam_param_tanh["diff_param"].item().items()}

Adam_drift_param_deep = {k: jnp.asarray(v) for k, v in Adam_param_deep["drift_param"].item().items()}
Adam_diff_param_deep  = {k: jnp.asarray(v) for k, v in Adam_param_deep["diff_param"].item().items()}
```

```python
# get true and ARFF data
T_drift = true_drift(x)
T_diffusion = true_diffusion(x)

ARFF_drift = ARFF_Functions.drift(ARFF_drift_param, x)
ARFF_diffusion = ARFF_Functions.diffusion(ARFF_diff_param, x, diff_type)

Adam_drift_FF = Adam_Functions_FF.drift(Adam_drift_param_FF, x)
Adam_diffusion_FF = Adam_Functions_FF.diffusion(Adam_diff_param_FF, x, diff_type)

Adam_drift_tanh = Adam_Functions_tanh.drift(Adam_drift_param_tanh, x)
Adam_diffusion_tanh = Adam_Functions_tanh.diffusion(Adam_diff_param_tanh, x, diff_type)

Adam_drift_deep = Adam_Functions_tanh.drift(Adam_drift_param_deep, x)
Adam_diffusion_deep = Adam_Functions_tanh.diffusion(Adam_diff_param_deep, x, diff_type)
```

```python
# RMSE data
ARFF_drift_RMSE = np.sqrt(np.mean((T_drift - ARFF_drift)**2))
ARFF_diff_RMSE = np.sqrt(np.mean((T_diffusion - ARFF_diffusion)**2)) 

Adam_drift_RMSE_FF = np.sqrt(np.mean((T_drift - Adam_drift_FF)**2))
Adam_diff_RMSE_FF = np.sqrt(np.mean((T_diffusion - Adam_diffusion_FF)**2)) 

Adam_drift_RMSE_tanh = np.sqrt(np.mean((T_drift - Adam_drift_tanh)**2))
Adam_diff_RMSE_tanh = np.sqrt(np.mean((T_diffusion - Adam_diffusion_tanh)**2)) 

Adam_drift_RMSE_deep = np.sqrt(np.mean((T_drift - Adam_drift_deep)**2))
Adam_diff_RMSE_deep = np.sqrt(np.mean((T_diffusion - Adam_diffusion_deep)**2)) 

print(ARFF_drift_RMSE, Adam_drift_RMSE_FF, Adam_drift_RMSE_tanh, Adam_drift_RMSE_deep)
print(ARFF_diff_RMSE, Adam_diff_RMSE_FF, Adam_diff_RMSE_tanh, Adam_diff_RMSE_deep)

# save RMSE
# np.savez(f"saved_results/true_trained_RMSE/{ex_name}_ARFF.npz", ARFF_drift_RMSE, ARFF_diff_RMSE)
# np.savez(f"saved_results/true_trained_RMSE/{ex_name}_Adam_FF.npz", Adam_drift_RMSE_FF, Adam_diff_RMSE_FF)
# np.savez(f"saved_results/true_trained_RMSE/{ex_name}_Adam_tanh.npz", Adam_drift_RMSE_tanh, Adam_diff_RMSE_tanh)
# np.savez(f"saved_results/true_trained_RMSE/{ex_name}_Adam_deep.npz", Adam_drift_RMSE_deep, Adam_diff_RMSE_deep)
```

```python
# format diffusion matrices
if diff_type == "diagonal":
    idx_map = {d: (d, d) for d in range(yD)}
    
    T_diffusion = np.diagonal(T_diffusion, axis1=1, axis2=2)
    ARFF_diffusion = np.diagonal(ARFF_diffusion, axis1=1, axis2=2)
    Adam_diffusion_FF = np.diagonal(Adam_diffusion_FF, axis1=1, axis2=2)
    Adam_diffusion_tanh = np.diagonal(Adam_diffusion_tanh, axis1=1, axis2=2)
    Adam_diffusion_deep = np.diagonal(Adam_diffusion_deep, axis1=1, axis2=2)
else:
    LT_rows, LT_cols = np.tril_indices(YD)
    LT_index_map = {(r, c): i for i, (r, c) in enumerate(zip(LT_rows, LT_cols))}
    idx_map = {i: (r, c) for i, (r, c) in enumerate(zip(LT_rows, LT_cols))}

    # Flatten to (N, LT)
    T_diffusion     = T_diffusion[:, LT_rows, LT_cols]
    ARFF_diffusion  = ARFF_diffusion[:, LT_rows, LT_cols]
    Adam_diffusion_FF  = Adam_diffusion_FF[:, LT_rows, LT_cols]
    Adam_diffusion_tanh  = Adam_diffusion_tanh[:, LT_rows, LT_cols]
    Adam_diffusion_deep  = Adam_diffusion_deep[:, LT_rows, LT_cols]
```

```python

def plotting_2D(fig, ax, data_list, d, name):
    """Plot a single row of 2D scatter plots with colorbar."""
    # Normalize across all models
    vmin = min(arr[:, d].min() for arr in data_list)
    vmax = max(arr[:, d].max() for arr in data_list)
    norm = Normalize(vmin=vmin, vmax=vmax)

    # Scatter plots
    for j, arr in enumerate(data_list[:-1]):
        ax[d, j].scatter(x[:, 0], x[:, 1], c=arr[:, d], cmap='viridis', s=20, norm=norm)
    sc = ax[d, -1].scatter(x[:, 0], x[:, 1], c=data_list[-1][:, d], cmap='viridis', s=20, norm=norm)

    # Colorbar
    bbox = ax[d, -1].get_position()
    cbar_ax = fig.add_axes([bbox.x1 + 0.01, bbox.y0, 0.015, bbox.height])
    fig.colorbar(sc, cax=cbar_ax).ax.set_ylabel(name, rotation=270, labelpad=20, fontsize=16)

    # Titles only for first row
    if d == 0:
        titles = ["True", "Algorithm 2", r"Adam $\mathcal{N}_K$", "Adam tanh Shallow", "Adam tanh Deep"]
        for j, title in enumerate(titles):
            ax[0, j].set_title(title, fontsize=13)


def plot_2D_grid(data_list, var_names, filename=None):
    yD = data_list[0].shape[1]
    fig, ax = plt.subplots(yD, 5, figsize=(20, 4*yD), gridspec_kw={'hspace':0.12, 'wspace':0.19})
    ax = np.atleast_2d(ax)
    for d in range(yD):
        plotting_2D(fig, ax, data_list, d, var_names[d])
    fig.supxlabel(r"Susceptible fraction $y_0$", fontsize=16, y=0.03)
    fig.supylabel(r"Recovered fraction $y_1$", fontsize=16, x=0.09)
    # fig.supxlabel(r"$v$", fontsize=16, y=-0.01)
    # fig.supylabel(r"$x$", fontsize=16, x=0.09)
    if filename is not None:
        fig.savefig(filename, dpi=300, bbox_inches='tight')
    plt.show()


# Drift
var_names_drift = [rf"$f_{{{d}}}(x)$" for d in range(T_drift.shape[1])]
plot_2D_grid([T_drift, ARFF_drift, Adam_drift_FF, Adam_drift_tanh, Adam_drift_deep], var_names_drift, 
             filename=f"saved_results/true_trained_plots/{ex_name}_drift.png")

# Diffusion
var_names_diff = [rf"$\sigma_{{{i},{j}}}(x)$" for (i,j) in idx_map.values()]
plot_2D_grid([T_diffusion, ARFF_diffusion, Adam_diffusion_FF, Adam_diffusion_tanh, Adam_diffusion_deep], var_names_diff,
             filename=f"saved_results/true_trained_plots/{ex_name}_diffusion.png")
```

```python

def plot_1D(Adam_drift, Adam_diffusion):
    fig, ax = plt.subplots(1, 2, figsize=(7, 3.55), constrained_layout=True)

    # plot true
    ax[0].plot(x, T_drift, color="black", linestyle="--", linewidth=1.0, label="True")
    ax[1].plot(x, T_diffusion.ravel(), color="black", linestyle="--", linewidth=1.0, label="True")

    # plot ARFF
    ax[0].plot(x, ARFF_drift, color="blue", label="ARFF")
    ax[1].plot(x, ARFF_diffusion.ravel(), color="blue", label="ARFF")

    # plot Adam
    ax[0].plot(x, Adam_drift, color="yellow", label="Adam")
    ax[1].plot(x, Adam_diffusion.ravel(), color="yellow", label="Adam")

    ax[0].set_xlabel("x")
    ax[0].set_ylabel("Drift")
    ax[0].legend()
    
    ax[1].set_xlabel("x")
    ax[1].set_ylabel("Diffusion")
    ax[1].legend()

def Plot_ND(Adam_drift, Adam_diffusion):
    if diff_type == "diagonal":
        # pull diag diffusion values
        T_diffusion_D = np.diagonal(T_diffusion, axis1=1, axis2=2)
        ARFF_diffusion_D = np.diagonal(ARFF_diffusion, axis1=1, axis2=2)
        Adam_diffusion_D = np.diagonal(Adam_diffusion, axis1=1, axis2=2)

        fig, ax = plt.subplots(YD, 2, figsize=(7, 3.55*YD), constrained_layout=True)
        ax = np.atleast_2d(ax)
        for d in range(YD):
            # Drift
            ax[d,0].scatter(T_drift[:,d], ARFF_drift[:,d], s=10, alpha=0.02, color='blue', label='ARFF')
            ax[d,0].scatter(T_drift[:,d], Adam_drift[:,d], s=10, alpha=0.02, color='yellow', label='Adam')
            min_val = min(T_drift[:,d].min(), ARFF_drift[:,d].min(), Adam_drift[:,d].min())
            max_val = max(T_drift[:,d].max(), ARFF_drift[:,d].max(), Adam_drift[:,d].max())
            ax[d,0].plot([min_val, max_val],[min_val,max_val],'k--', lw=1)
            ax[d,0].set_aspect('equal', adjustable='box')
            
            # Diffusion
            ax[d,1].scatter(T_diffusion_D[:,d], ARFF_diffusion_D[:,d], s=10, alpha=0.02, color='blue', label='ARFF')
            ax[d,1].scatter(T_diffusion_D[:,d], Adam_diffusion_D[:,d], s=10, alpha=0.02, color='yellow', label='Adam')
            min_val = min(T_diffusion_D[:,d].min(), ARFF_diffusion_D[:,d].min(), Adam_diffusion_D[:,d].min())
            max_val = max(T_diffusion_D[:,d].max(), ARFF_diffusion_D[:,d].max(), Adam_diffusion_D[:,d].max())
            ax[d,1].plot([min_val, max_val],[min_val,max_val],'k--', lw=1)
            ax[d,1].set_aspect('equal', adjustable='box')

    else:
        LT_rows, LT_cols = np.tril_indices(YD)
        LT_index_map = {(r, c): i for i, (r, c) in enumerate(zip(LT_rows, LT_cols))}

        # Flatten to (N, LT)
        T_diffusion_LT     = T_diffusion[:, LT_rows, LT_cols]
        ARFF_diffusion_LT  = ARFF_diffusion[:, LT_rows, LT_cols]
        Adam_diffusion_LT  = Adam_diffusion[:, LT_rows, LT_cols]
        
        fig, ax = plt.subplots(YD, YD+1, figsize=(3.5*(YD+1), 3.55*YD), constrained_layout=True)
        ax = np.atleast_2d(ax)
        
        for d in range(YD):
            # Drift column
            ax[d,0].scatter(T_drift[:,d], ARFF_drift[:,d], s=10, alpha=0.01, color='blue', label='ARFF')
            ax[d,0].scatter(T_drift[:,d], Adam_drift[:,d], s=10, alpha=0.01, color='yellow', label='Adam')
            min_val = min(T_drift[:,d].min(), ARFF_drift[:,d].min(), Adam_drift[:,d].min())
            max_val = max(T_drift[:,d].max(), ARFF_drift[:,d].max(), Adam_drift[:,d].max())
            ax[d,0].plot([min_val,max_val],[min_val,max_val],'k--', lw=1)
            ax[d,0].set_title(r"$f_{" + f"{d}" + r"}$")
            ax[d,0].set_aspect('equal', adjustable='box')
            
            # Diffusion columns (D columns for D×D)
            for col in range(YD):
                if d >= col:  # lower-triangular
                    idx = LT_index_map[(d,col)]
                    ax[d,col+1].scatter(T_diffusion_LT[:,idx], ARFF_diffusion_LT[:,idx], s=10, alpha=0.01, color='blue')
                    ax[d,col+1].scatter(T_diffusion_LT[:,idx], Adam_diffusion_LT[:,idx], s=10, alpha=0.01, color='yellow')
                    min_val = min(T_diffusion_LT[:,idx].min(), ARFF_diffusion_LT[:,idx].min(), Adam_diffusion_LT[:,idx].min())
                    max_val = max(T_diffusion_LT[:,idx].max(), ARFF_diffusion_LT[:,idx].max(), Adam_diffusion_LT[:,idx].max())
                    ax[d,col+1].plot([min_val,max_val],[min_val,max_val],'k--', lw=1)
                    ax[d, col+1].set_title(r"$\sigma_{" + f"{d},{col}" + r"}$")
                    ax[d,col+1].set_aspect('equal', adjustable='box')
                else:
                    ax[d,col+1].axis('off')  # blank upper-triangle
                    
    handles = [plt.Line2D([0],[0], marker='o', color='w', markerfacecolor='blue', markersize=6),
               plt.Line2D([0],[0], marker='o', color='w', markerfacecolor='yellow', markersize=6)]
    labels = ['Algorithm 2','Adam']

    for d in range(YD):
        for col in range(YD + 1):
            a = ax[d, col]
            a.xaxis.set_major_locator(mticker.MaxNLocator(nbins=5))
            a.yaxis.set_major_locator(mticker.MaxNLocator(nbins=5))
            # Only add legend if this axis is not an upper-triangular blank
            if col == 0 or d >= (col - 1):
                a.legend(handles, labels, loc="best")
            
    fig.supxlabel("True", fontsize=14)
    fig.supylabel("Trained", fontsize=14)

#fig.savefig(f"saved_results/true_trained_plots/{ex_name}{Adam_type}.png", dpi=300, bbox_inches='tight')#, pad_inches=0.05)
```
