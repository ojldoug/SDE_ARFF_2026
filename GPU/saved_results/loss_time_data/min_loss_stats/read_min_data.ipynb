```python
import numpy as np
import os

for fname in sorted(os.listdir(".")):
    if not fname.endswith(".npz"):
        continue

    data = np.load(fname)

    if fname.endswith("ARFF.npz"):
        # Case 1: ARFF-format files with T_mean and L_mean
        L_mean = data["L_mean"].item() if data["L_mean"].ndim == 0 else data["L_mean"]

        print(f"{fname}: Mean min loss = {L_mean:.3f}")

    else:
        # Case 2: simple npz with a single float inside
        # Extract the first (and only) key
        key = list(data.keys())[0]
        val = data[key].item() if np.ndim(data[key]) == 0 else data[key]

        print(f"{fname}: Mean min loss = {val:.3f}")
```
