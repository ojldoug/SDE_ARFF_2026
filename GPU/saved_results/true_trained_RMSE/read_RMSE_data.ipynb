```python
import os
import numpy as np

for fname in sorted(os.listdir(".")):
    if fname.endswith(".npz"):
        data = np.load(fname)
        # unnamed keys when saved without keywords
        drift_rmse = data[data.files[0]]
        diff_rmse = data[data.files[1]]
        print(f"{fname}: drift = {drift_rmse}, diffusion = {diff_rmse}")

```
