```python
import os
import dill
import jax.numpy as jnp
import numpy as np
```

```python
# ex1 2d cubic
def true_drift(x):
    def f_1(x):
        A = -1.5
        B = jnp.array([[8], [0]])
        C = jnp.array([[0], [0]])
        D = jnp.array([[-16], [0]])
        f = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return f
    
    def f_2(x):
        A = -1.5
        B = jnp.array([[0], [8]])
        C = jnp.array([[0], [0]])
        D = jnp.array([[0], [-16]])
        f = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return f

    drift = jnp.transpose(jnp.array([f_1(x), f_2(x)]))
    return drift


def true_diffusion(x):
    def sigma_11(x):
        A = 0.5  
        B = jnp.array([[0.1], [0]])
        C = jnp.array([[0], [0]])
        D = jnp.array([[0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma
    
    def sigma_12(x):
        A = 0
        B = jnp.array([[0], [0]])
        C = jnp.array([[0], [0]])
        D = jnp.array([[0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma
    
    def sigma_22(x):
        A = 0.5
        B = jnp.array([[0], [0.1]])
        C = jnp.array([[0], [0]])
        D = jnp.array([[0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    diffusion = jnp.abs(jnp.transpose(jnp.array([[sigma_11(x), sigma_12(x)], [sigma_12(x), sigma_22(x)]])))
    return diffusion

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "n_dimensions": 2,
    "diff_type": "diagonal",
    "xlim": np.array([[-1, 1], [-1, 1]]),
}

with open("ex1.pkl", "wb") as f:
    dill.dump(functions, f)
```

```python
# ex2 LT
def true_drift(x):
    def f_1(x):
        A = 0
        B = jnp.array([[-1], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        f = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return f

    def f_2(x):
        A = 0
        B = jnp.array([[0], [-1], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        f = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return f

    def f_3(x):
        A = 0
        B = jnp.array([[0], [0], [-1]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        f = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return f

    drift = jnp.transpose(jnp.array([f_1(x), f_2(x), f_3(x)]))
    return drift


def true_diffusion(x):
    def sigma_11(x):
        A = 0.09506174
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_12(x):
        A = 0.04639143
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_13(x):
        A = 0.04337843
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_22(x):
        A = 0.15817465
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_23(x):
        A = 0.07506578
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_33(x):
        A = 0.00852886
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_21(x):
        A = 0
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_31(x):
        A = 0
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    def sigma_32(x):
        A = 0
        B = jnp.array([[0], [0], [0]])
        C = jnp.array([[0], [0], [0]])
        D = jnp.array([[0], [0], [0]])
        sigma = jnp.squeeze(A + jnp.dot(x, B) + jnp.dot(jnp.square(x), C) + jnp.dot(jnp.power(x, 3), D))
        return sigma

    diffusion = jnp.transpose(jnp.array(
        [[sigma_11(x), sigma_12(x), sigma_13(x)], 
         [sigma_21(x), sigma_22(x), sigma_23(x)],
         [sigma_31(x), sigma_32(x), sigma_33(x)]]))
    return diffusion

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "n_dimensions": 3,
    "diff_type": "triangular",
    "xlim": jnp.array([[-1, 1], [-1, 1], [-1, 1]]),
}

with open("ex2.pkl", "wb") as f:
    dill.dump(functions, f)
```

```python
# ex4 Langevin
xlim = np.array([[-1, 1], [-1, 1]])

def true_drift(x):
    v, x = x[:,0], x[:,1]
    drift = (-x**3 - x - 0.5 * v)
    return drift[:, None]

def true_diffusion(x):
    diff = jnp.sqrt(0.1)
    return jnp.tile(diff, (x.shape[0], 1, 1))

def coupled_fn(y0, x0, h):
    y0_flat = jnp.ravel(y0)  # or y0[:,0]
    return x0.at[:, 1].add(y0_flat * h)

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "coupled_fn": coupled_fn,
    "n_dimensions": 1,
    "diff_type": "diagonal",
    "xlim": jnp.array([[-1, 1], [-1, 1]]),
}

with open("ex4.pkl", "wb") as f:
    dill.dump(functions, f)
```

```python
# ex5 Gillespie
def true_drift(x):
    k1, k2 = 1, 1
    theta0 = jnp.clip(x[:, 0], 0, 1)
    theta2 = jnp.clip(x[:, 1], 0, 1 - theta0)
    theta1 = jnp.clip(1 - theta0 - theta2, 0, 1)
    r1 = 4 * k1 * theta0 * theta1
    r2 = k2 * theta1
    return jnp.stack([-r1, r2], axis=1)

def true_diffusion(x):
    N = 1024
    k1, k2 = 1, 1
    theta0 = jnp.clip(x[:, 0], 0, 1)
    theta2 = jnp.clip(x[:, 1], 0, 1 - theta0)
    theta1 = jnp.clip(1 - theta0 - theta2, 0, 1)
    r1 = 4 * k1 * theta0 * theta1
    r2 = k2 * theta1
    epsilon = 1e-8
    # construct diagonal diffusion for each sample
    diag0 = jnp.sqrt(r1 / N + epsilon)
    diag1 = jnp.sqrt(r2 / N + epsilon)
    zeros = jnp.zeros_like(diag0)
    # shape (batch, 2, 2)
    diffusion = jnp.stack([jnp.stack([diag0, zeros], axis=1),
                           jnp.stack([zeros, diag1], axis=1)], axis=1)
    return diffusion

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "n_dimensions": 2,
    "diff_type": "diagonal",
}

with open("ex5.pkl", "wb") as f:
    dill.dump(functions, f)
```

```python
# ex6 SPDE 
def true_drift(x):
    x = jnp.atleast_2d(x)
    x = x[:, 1]
    return (jnp.sin(4 * jnp.pi * x) * 5).reshape(-1, 1)

def true_diffusion(x):
    space_size = 1
    x = jnp.atleast_2d(x)
    x = x[:, 1]
    diffusion = 0.05 * (1 + jnp.exp(-(x - space_size / 2) ** 2 * 150))
    SPDE_diffusion = diffusion * 0.5  # corrective term
    return SPDE_diffusion.reshape(-1, 1, 1)

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "n_dimensions": 1,
    "diff_type": "diagonal",
}

with open("ex6.pkl", "wb") as f:
    dill.dump(functions, f)
```

```python
# ex7 2D fat tail
def true_drift(x):
    x = jnp.atleast_2d(x)   # shape (N,2)
    x1, x2 = x[:,0], x[:,1]
    a = 0.1
    b = 0.2
    f1 = jnp.exp(-jnp.abs(x1)/a) * jnp.exp(-0.5*(x1**2 + x2**2))
    f2 = jnp.exp(-jnp.abs(x2)/b) * jnp.exp(-0.5*(x1**2 + x2**2))

    drift = jnp.stack([f1, f2], axis=1)   # shape (N,2)
    return drift

def true_diffusion(x):
    x = jnp.atleast_2d(x)
    diff = 0.1

    # scalar diffusion but expanded to isotropic 2x2 matrices
    N = x.shape[0]
    diffusion = diff * jnp.tile(jnp.eye(2), (N,1,1))   # shape (N,2,2)
    return diffusion

functions = {
    "drift": true_drift,
    "diffusion": true_diffusion,
    "n_dimensions": 2,
    "diff_type": "diagonal",
    "xlim": np.array([[-1, 1], [-1, 1]]),
}

with open("ex7.pkl", "wb") as f:
    dill.dump(functions, f)
```
